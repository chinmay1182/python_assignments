## 2. Write a Python program to read first n lines of a file.

```python
n = int(input("Enter number of lines to read: "))

with open("Text.txt.txt", "r") as file:
    for i in range(n):
        line = file.readline()
        if line == "":
            break
        print(line, end="")
```

## 3. Write a Python program to append text to a file and display the text.

```python
# append text to file
with open("Text.txt.txt", "a") as file:
    text = input("Enter text to append: ")
    file.write(text + "\n")

# display file content
with open("Text.txt.txt", "r") as file:
    print("\nFile Content:")
    print(file.read())
```

## 4. Write a Python program to read last n lines of a file.

```python
n = int(input("Enter number of last lines to read: "))

with open("Text.txt.txt", "r") as file:
    lines = file.readlines()
    last_lines = lines[-n:]

for line in last_lines:
    print(line, end="")
```

## 5. Write a Python program to read a file line by line and store it into a list.

```python
lines_list = []

with open("Text.txt.txt", "r") as file:
    for line in file:
        lines_list.append(line.strip())

print("Lines stored in list:")
print(lines_list)
```

## 6. Write a Python program to read a file line by line store it into a variable.

```python
content = ""

with open("Text.txt.txt", "r") as file:
    for line in file:
        content += line

print("File Content:")
print(content)
```

## 7. Write a python program to find the longest words.

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

max_len = max(len(word) for word in words)

print("Longest word(s):")
for word in words:
    if len(word) == max_len:
        print(word)
```

## 8. Write a Python program to count the number of lines in a text file.

```python
count = 0

with open("Text.txt.txt", "r") as file:
    for line in file:
        count += 1

print("Number of lines in file:", count)
```

## 9. Write a Python program to count the frequency of words in a file.

```python
with open("Text.txt.txt", "r") as file:
    text = file.read()

words = text.split()
word_freq = {}

for word in words:
    if word in word_freq:
        word_freq[word] += 1
    else:
        word_freq[word] = 1

print("Word Frequency:")
for word, count in word_freq.items():
    print(word, ":", count)
```
