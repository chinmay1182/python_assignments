1. Write a program to add two lists index-wise. Create a new list that contains the 0th index item from both the list,
then the 1st index item, and so on till the last element. any leftover items will get added at the end of the new list.

```python
# Given List:
# list1 = ["M", "na", "i", "Ra"]
# list2 = ["y", "me", "s", "hul"]
# Output:
# [['M','y'], ['na', me'], ['i', 's'], ['Ra', 'hul']]

list1 = ["M", "na", "i", "Ra"]
list2 = ["y", "me", "s", "hul"]

result = []

for i in range(len(list1)):
    result.append([list1[i], list2[i]])

print(result)
```

7. Sort a list of alphanumeric strings based on product value of numeric character in it.
     If in any string there is no numeric character take it's product value as 1.

```python
# Input:
# ['1ac21', '23fg', '456', '098d','1','kls']
# Output:
# ['456', '23fg', '1ac21', '1', 'kls', '098d']

lst = ['1ac21', '23fg', '456', '098d','1','kls']

def product_value(s):
    product = 1
    found = False

    for ch in s:
        if ch.isdigit():
            product *= int(ch)
            found = True

    if not found:
        return 1
    return product

result = sorted(lst, key=product_value, reverse=True)

print(result)
```

16. Write a Python function to check whether a number is perfect or not.

```python
# A Perfect number is a number that is half the sum of all of its positive divisors (including itself).
#
# Example :
# The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
# Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.
#
# The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.

def is_perfect(n):
    if n <= 0:
        return False
    divisor_sum = 0
    for i in range(1, n):
        if n % i == 0:
            divisor_sum += i
    return divisor_sum == n

def check_perfect(num):
    sum = 0

    for i in range(1, num):
        if num % i == 0:
            sum = sum + i

    if sum == num:
        print(num, "is a Perfect Number")
    else:
        print(num, "is not a Perfect Number")

n = int(input("Enter a number: "))
check_perfect(n)
```

17. Write a Python function to concatenate any no of dictionaries to create a new one.

```python
# Sample Dictionary :
# dic1={1:10, 2:20}
# dic2={3:30, 4:40}
# dic3={5:50,6:60}
# Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}

def concat_dict(*dicts):
    result = {}

    for d in dicts:
        result.update(d)

    return result

dic1 = {1:10, 2:20}
dic2 = {3:30, 4:40}
dic3 = {5:50, 6:60}

print(concat_dict(dic1, dic2, dic3))
```

18. Write a python function that receives a list of integers and prints out a histogram of bin size 10

```python
# Input:
# [13,42,15,37,22,39,41,50]
# Output:
# {11-20:2,21-30:1,31-40:2,41-50:3}

def print_histogram(nums):
    if not nums:
        print("{}")
        return

    min_val = min(nums)
    max_val = max(nums)

    bins_start = ((min_val - 1) // 10) * 10 + 1
    bins_end = ((max_val + 9) // 10) * 10

    hist = {}

    for start in range(bins_start, bins_end + 1, 10):
        end = start + 9
        count = sum(1 for x in nums if start <= x <= end)

        if count > 0:
            hist[f"{start}-{end}"] = count

    print(hist)

nums = [13,42,15,37,22,39,41,50]
print_histogram(nums)
```
